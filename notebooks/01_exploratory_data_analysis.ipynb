```python
#Bag Ratio Features
dataset_conv_all["bag_ratio"] = (dataset_conv_all["total_bags"]/dataset_conv_all["total_volume"].replace(0,np.nan))
dataset_conv_all["small_bag_ratio"] = (dataset_conv_all["small_bags"]/ dataset_conv_all["total_bags"].replace(0, np.nan))
dataset_conv_all["large_bag_ratio"] = (dataset_conv_all["large_bags"]/ dataset_conv_all["total_bags"].replace(0, np.nan))
dataset_conv_all["xlarge_bag_ratio"] = (dataset_conv_all["xlarge_bags"]/ dataset_conv_all["total_bags"].replace(0, np.nan))
```

```python
##PLU Ratio featire
dataset_conv_all["PLU4046_ratio"] = (dataset_conv_all["4046"]/ dataset_conv_all["total_volume"].replace(0, np.nan))
dataset_conv_all["PLU4225_ratio"] = (dataset_conv_all["4225"]/ dataset_conv_all["total_volume"].replace(0, np.nan))
dataset_conv_all["PLU4770_ratio"] = (dataset_conv_all["4770"]/ dataset_conv_all["total_volume"].replace(0, np.nan))
```

```python
Relative_Features = ["average_price", "bag_ratio", "small_bag_ratio",    "large_bag_ratio",    "xlarge_bag_ratio",    "PLU4046_ratio",    "PLU4225_ratio",    "PLU4770_ratio"]

dataset_conv_all[Relative_Features].corr()
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(dataset_conv_all[Relative_Features].corr(),annot=True,fmt=".2f",cmap="coolwarm",center=0)
plt.title("Korrelation relativer Mengen-Features mit dem Preis")
plt.show()
```
